```python
import numpy as np
import matplotlib.pyplot as plt

cats = ['A', 'B', 'C', 'D']
v1, v2, v3, v4 =[4, 7, 1, 8], [5, 6, 2, 9], [2, 6, 4, 3], [4, 2, 3, 6]
w, x = 0.2, np.arange(len(cats))

# Shift each bar by -1.5w, -0.5w, 0.5w, and 1.5w to center the group of 4 bars
plt.bar(x - 1.5 * w, v1, w, label='Set 1')
plt.bar(x - 0.5 * w, v2, w, label='Set 2')
plt.bar(x + 0.5 * w, v3, w, label='Set 3')
plt.bar(x + 1.5 * w, v4, w, label='Set 4')

plt.xticks(x, cats)
plt.ylabel('Values')
plt.title('Grouped Bar Chart')
plt.legend()
plt.show()
```
